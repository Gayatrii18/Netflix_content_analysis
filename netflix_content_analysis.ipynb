```python
df['director'] = df['director'].fillna('unknown')
ddtf['cast'] = df['cast'].fillna('unknown')
df['country'] = df['country'].fillna('unknown')
```

```python
df.dropna(subset=['date_added', 'rating'], inplace=True)
```

```python
df.isnull().sum()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'])
df['year_added'] = df['date_added'].dt.year
df[['title', 'date_added', 'year_added']].head()
```

```python
growth_data = df['year_added'].value_counts().sort_index()
```

```python
growth_data
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(x=growth_data.index, y=growth_data.values, marker='o', color='#ff0000')

plt.title('Netflix Content Addition Growth (2008 - 2021)', fontsize=15)
plt.xlabel('Year Added')
plt.ylabel('Number of Movies/TV Shows')
plt.grid(True)
plt.show()
```

```python
type_growth = df.groupby(['year_added', 'type'])['show_id'].count().reset_index()
type_growth.columns = ['Year', 'Type', 'Count']

plt.figure(figsize=(12, 6))
sns.lineplot(data=type_growth, x='Year', y='Count', hue='Type', marker='o', palette=['#ff0000', '#0000ff'])

plt.title('Movies vs TV Shows Added Over the Years', fontsize=16)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
df['country_list'] = df['country'].str.split(', ')
```

```python
exploded_countries = df.explode('country_list')
```

```python
top_countries = exploded_countries['country_list'].value_counts().head(10)
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 Content Producing Countries (Accurate Count)')
plt.show()
```

```python
tv_shows = df[df['type'] == 'TV Show'].copy()

tv_shows['seasons'] = tv_shows['duration'].str.extract('(\d+)').astype(int)

sns.histplot(tv_shows['seasons'], bins=20, kde=False, color='blue')
plt.title('Distribution of TV Show Seasons')
plt.xlabel('Number of Seasons')
plt.ylabel('Count')
plt.show()
```
